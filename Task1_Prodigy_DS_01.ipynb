Create a bar chart or histogram to visualize the distribution of a categorical or continuous variable.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("task1 dataset.csv", skiprows=4)

df.head()
```

## Dataset Description

The dataset is obtained from the World Development Indicators (WDI) database. It contains population statistics for different countries from 1960 to 2025.

For this task, the population values for the year 2025 are used for visualization.

```python
print(df.columns)
```

```python
df.info()
```

## Histogram of Population Distribution (2025)

```python
plt.figure(figsize=(10,6))

plt.hist(df["2025"].dropna(), bins=20, edgecolor="black",color="skyblue")

plt.title("Distribution of Population (2025)")
plt.xlabel("Population")
plt.ylabel("Number of Countries")

plt.grid(True)

plt.show()
```

### Insights

• Most countries have relatively small populations.

• Only a few countries have extremely large populations.

• The histogram is positively skewed because countries like India and China have much larger populations than most other countries.

• This visualization helps understand how population is distributed globally.

## Top 10 Most Populous Countries (2025)

```python
top10 = df.sort_values("2025", ascending=False).head(10)

plt.figure(figsize=(12,6))

plt.bar(top10["Country Name"],top10["2025"],color="skyblue")

plt.xticks(rotation=45)

plt.title("Top 10 Most Populous Countries (2025)")
plt.xlabel("Country")
plt.ylabel("Population")

plt.show()
```

### Insights

• India has the highest population in 2025.

• China is the second most populous country.

• There is a noticeable difference between the top two countries and the remaining countries.

• The bar chart makes it easy to compare the populations of the top 10 countries.

# Conclusion

The histogram provides an overview of the overall population distribution, while the bar chart highlights the population of the top 10 countries.

These visualizations make it easier to understand trends and compare population values across countries.
